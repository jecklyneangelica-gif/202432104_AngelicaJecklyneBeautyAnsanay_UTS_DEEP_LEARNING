  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>
</div>

```python
#202432104_AngelicaJecklyneBeautyAnsanay
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

torch.manual_seed(42)
np.random.seed(42)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432104_AngelicaJecklyneBeautyAnsanay
# 1. Muat dataset (ubah ke URL raw GitHub jika dijalankan dari repositori)
DATA_PATH = 'ispu_dki1.csv'
df = pd.read_csv(DATA_PATH, parse_dates=['tanggal'])
print('Ukuran dataset   :', df.shape)
print('Rentang tanggal  :', df['tanggal'].min().date(), 's.d.', df['tanggal'].max().date())
print('PM10 kosong      :', df['pm10'].isna().sum())

# 2. Pembersihan & pra-pemrosesan (target: PM10 harian)
series = df.sort_values('tanggal').set_index('tanggal')['pm10']
series = series.reindex(pd.date_range(series.index.min(), series.index.max(), freq='D'))
series = series.loc[:'2021-12-31']
print('Hari tanpa data 2010-2021:', series.isna().sum())
series = series.interpolate(method='linear', limit_direction='both')

window = 30
train_size = int(len(series) * 0.8)
data_min = series.iloc[:train_size].min()
data_max = series.iloc[:train_size].max()
scaled = ((series - data_min) / (data_max - data_min)).values.astype('float32')

def create_sequences(data, window):
    X, y = [], []
    for i in range(len(data) - window):
        X.append(data[i:i + window])
        y.append(data[i + window])
    return np.array(X)[..., np.newaxis], np.array(y)[:, np.newaxis]

X, y = create_sequences(scaled, window)

# 3. Bagi data secara kronologis dan konversi ke PyTorch Tensor
split = train_size - window
X_train, y_train = torch.tensor(X[:split]), torch.tensor(y[:split])
X_test, y_test = torch.tensor(X[split:]), torch.tensor(y[split:])
test_dates = series.index[train_size:]
print('X_train:', tuple(X_train.shape), '| X_test:', tuple(X_test.shape))
print('Periode uji      :', test_dates[0].date(), 's.d.', test_dates[-1].date())
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432104_AngelicaJecklyneBeautyAnsanay
class SequenceModel(nn.Module):
    def __init__(self, input_size=1, hidden_size=64, output_size=1):
        super(SequenceModel, self).__init__()
        self.rnn = nn.RNN(input_size, hidden_size, batch_first=True)
        self.fc = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        out, h_n = self.rnn(x)
        out = out[:, -1, :]
        return self.fc(out)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432104_AngelicaJecklyneBeautyAnsanay
# 1. Inisiasi hyperparameter
epochs = 50
learning_rate = 0.01
batch_size = 64
hidden_size = 64

# 2. Inisiasi model, fungsi loss, optimizer, dan data loader
model = SequenceModel(hidden_size=hidden_size)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)
train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=batch_size, shuffle=True)

# 3. Simpan riwayat loss
train_losses = []
test_losses = []

# 4. Training loop
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    for xb, yb in train_loader:
        optimizer.zero_grad()
        loss = criterion(model(xb), yb)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        running_loss += loss.item() * xb.size(0)
    train_losses.append(running_loss / len(X_train))

    model.eval()
    with torch.no_grad():
        test_losses.append(criterion(model(X_test), y_test).item())

    if (epoch + 1) % 10 == 0:
        print(f'Epoch [{epoch + 1}/{epochs}] | Train Loss: {train_losses[-1]:.5f} | Test Loss: {test_losses[-1]:.5f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432104_AngelicaJecklyneBeautyAnsanay
# 1. Plot Grafik Penurunan Loss Training terhadap Epoch
plt.figure(figsize=(10, 4))
plt.plot(train_losses, color='#114F6A', linewidth=2, label='Training Loss')
plt.plot(test_losses, color='#1FA4B1', linewidth=2, linestyle='--', label='Test Loss')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE, data ternormalisasi)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# 2. Prediksi pada data uji dan kembalikan ke skala asli
model.eval()
with torch.no_grad():
    pred_scaled = model(X_test).numpy().flatten()
scale = data_max - data_min
pred = pred_scaled * scale + data_min
actual = y_test.numpy().flatten() * scale + data_min
naive = X_test[:, -1, 0].numpy() * scale + data_min

def metrics(a, p):
    mae = np.mean(np.abs(a - p))
    rmse = np.sqrt(np.mean((a - p) ** 2))
    r2 = 1 - np.sum((a - p) ** 2) / np.sum((a - a.mean()) ** 2)
    return mae, rmse, r2

for nama, p in [('RNN', pred), ('Baseline (nilai kemarin)', naive)]:
    mae, rmse, r2 = metrics(actual, p)
    print(f'{nama:<26} MAE = {mae:.2f} | RMSE = {rmse:.2f} | R2 = {r2:.3f}')

# 3. Plot Grafik Perbandingan Data Aktual vs Hasil Prediksi Model
plt.figure(figsize=(12, 5))
plt.plot(test_dates, actual, color='#114F6A', linewidth=1.2, label='Data Aktual')
plt.plot(test_dates, pred, color='#1FA4B1', linewidth=1.2, label='Prediksi RNN')
plt.title('Data Aktual vs Prediksi RNN pada Data Uji (PM10 Harian, DKI 1 Bundaran HI)')
plt.xlabel('Tanggal')
plt.ylabel('Indeks PM10 (ISPU)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# 4. Zoom 120 hari terakhir agar pola tiap hari terlihat jelas
plt.figure(figsize=(12, 4))
plt.plot(test_dates[-120:], actual[-120:], color='#114F6A', linewidth=1.8, marker='o', markersize=3, label='Data Aktual')
plt.plot(test_dates[-120:], pred[-120:], color='#1FA4B1', linewidth=1.8, marker='o', markersize=3, label='Prediksi RNN')
plt.title('Zoom 120 Hari Terakhir Data Uji')
plt.xlabel('Tanggal')
plt.ylabel('Indeks PM10 (ISPU)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```
